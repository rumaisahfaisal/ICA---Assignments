ride-hailing in pakistan

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_style('whitegrid')

df_ride = pd.read_csv('ride_hailing_dataset.csv')
print(df_ride.head())
```

```python
print(df_ride.groupby('service_type')['fare'].mean())
```

q1: average fare by service type is: inDrive pkr 573.51, careem pkr 544.98, rickshaw pkr 486.64 and uber pkr 486.13.

```python
print(df_ride['city'].value_counts())
```

q2: islamabad has the most trips with 29, then lahore (26), karachi (23) and rawalpindi (22).

```python
print(df_ride['distance_km'].mean())
```

q3: the average distance per trip is about 9.61 km.

```python
print(df_ride.groupby('gender')['trip_rating'].mean())
```

q4: male riders give a higher average rating (3.33) than female riders (2.82).

```python
avg_fare = df_ride.groupby('service_type')['fare'].mean().reset_index()

plt.figure(figsize=(6, 4))
sns.barplot(data=avg_fare, x='service_type', y='fare', hue='service_type', legend=False)
plt.title('average fare by service type')
plt.xlabel('service type')
plt.ylabel('average fare (pkr)')
plt.show()
```

```python
q5: inDrive has the highest average fare and uber has the lowest. the bar chart shows this.
```
