
# descriptive analytics – two averages, one argument

used in this notebook: the cleaned teaching prices dataset from week 2 to calculate the mean, median and first-to-last percentage change for each item. 

it also compares the mean and median of one item with and without its flagged value.

```python
import pandas as pd

df = pd.read_csv("../../teaching_prices_clean.csv")

print("dataset shape:", df.shape)
display(df.head())
```

```python
stats = df.groupby("item")["price"].agg(["mean", "median"])

print(stats.round(2))
```

```python
df["date"] = pd.to_datetime(df["date"])

daily = df.groupby(["item", "date"])["price"].mean().reset_index()

daily = daily.sort_values(["item", "date"])

change = daily.groupby("item")["price"].agg(["first", "last"])

change["percentage_change"] = (
    (change["last"] - change["first"]) / change["first"]
) * 100

print(change.round(2))
```

```python
sweetener = df[df["item"] == "Sweetener D"].copy()

print("without the flagged observation:")
print("mean:", sweetener["price"].mean())
print("median:", sweetener["price"].median())

sweetener_with_flag = sweetener.copy()

sweetener_with_flag.loc[
    (sweetener_with_flag["date"] == pd.Timestamp("2026-01-22")) &
    (sweetener_with_flag["city"] == "City 1"),
    "price"
] = 9999

print("\nwith the flagged observation:")
print("mean:", sweetener_with_flag["price"].mean())
print("median:", sweetener_with_flag["price"].median())
```
