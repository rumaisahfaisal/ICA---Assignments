```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
crop_df = pd.read_csv('crop_yield.csv')
```

```python
print(crop_df.head(10))
```

```python
print(crop_df.isnull().sum())
```

```python
print(crop_df.dtypes)
```

```python
print(crop_df.describe())
```

```python
print(crop_df.describe(include='object'))
```

```python
crop_df = crop_df.dropna(subset=['YIELD_KG'])
```

```python
crop_df['AREA_ACRES'] = crop_df['AREA_ACRES'].fillna(crop_df['AREA_ACRES'].mean())
```

```python
crop_df['DISTRICT'] = crop_df['DISTRICT'].fillna('Unknown')
```

```python
crop_df['FERTILIZER_USED'] = crop_df['FERTILIZER_USED'].fillna('No')
```

```python
crop_df['YIELD_VALUE'] = crop_df['YIELD_VALUE'].fillna(
    crop_df['YIELD_VALUE'].mean()
)
```

```python
crop_df['HARVEST_DATE'] = pd.to_datetime(
    crop_df['HARVEST_DATE'], errors='coerce', format='%d/%m/%Y'
)
```

```python
crop_df['FARM_ID'] = crop_df['FARM_ID'].str.replace('F', '').astype(int)
```

```python
print(crop_df.isnull().sum())
```

```python
print(crop_df.dtypes)
```

```python
crop_yield_total = crop_df.groupby('CROP_TYPE').agg({'YIELD_KG': 'sum'})
print(crop_yield_total)
```

```python
crop_yield_total.plot(kind='bar')
plt.title('Total Yield by Crop Type (KG)')
plt.ylabel('Yield (KG)')
plt.show()
```

```python
yield_pivot = crop_df.pivot_table(
    values='YIELD_KG', index='HARVEST_DATE', aggfunc='sum'
)
print(yield_pivot)
```

```python
yield_pivot.plot(kind='line')
plt.title('Yield Trend Over Time (KG)')
plt.ylabel('Yield (KG)')
plt.savefig('crop_yield_trend.png')
plt.show()
```
