student tea and snack preferences

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_style('whitegrid')

df_tea = pd.read_csv('student_tea_snacks.csv')
print(df_tea.head())
```

```python
print(df_tea['tea_type'].value_counts())
```

q1: green tea is the most preferred tea type, with 29 students. doodh patti has 25 and black has 23. 23 rows have no tea type listed, so they are not counted.

```python
print(df_tea.groupby('gender')['cups_per_day'].mean())
```

q2: males drink 3.0 cups per day on average and females drink 2.6 cups per day.

```python
snack_counts = df_tea['snack_choice'].value_counts().reset_index()

plt.figure(figsize=(6, 4))
sns.barplot(data=snack_counts, x='snack_choice', y='count', hue='snack_choice', legend=False)
plt.title('snack choices')
plt.xlabel('snack')
plt.ylabel('number of students')
plt.show()
```

q3: samosa is the most chosen snack (40 students), then chips (31) and biscuit (29).

```python
print(pd.crosstab(df_tea['age_group'], df_tea['prefers_tea']))

pd.crosstab(df_tea['age_group'], df_tea['prefers_tea']).plot(kind='bar')
plt.title('tea preference by age group')
plt.xlabel('age group')
plt.ylabel('number of students')
plt.show()
```

q4: in 18-20, 20 students prefer tea and 19 do not. in 21-23, 15 prefer tea and 17 do not. in 24-26, 16 prefer tea and 13 do not. the 24-26 group has the highest share of tea lovers, but the differences between the groups are small.
